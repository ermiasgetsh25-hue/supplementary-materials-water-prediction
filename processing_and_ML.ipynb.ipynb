```python
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

# Suppress the specific warnings about unknown categories and joblib parallel
warnings.filterwarnings("ignore", category=UserWarning, module="sklearn.preprocessing._encoders")
warnings.filterwarnings("ignore", category=UserWarning, module="sklearn.utils.parallel")

from sklearn.base import BaseEstimator, TransformerMixin, ClassifierMixin
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.feature_selection import SelectFromModel
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score
)
from sklearn.utils.class_weight import compute_class_weight
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE
from imblearn.combine import SMOTEENN
from statsmodels.stats.outliers_influence import variance_inflation_factor
from collections import Counter
import joblib

# ============================================================
# 1. CUSTOM TRANSFORMERS
# ============================================================

class CorrelationVIFSelector(BaseEstimator, TransformerMixin):
    """Removes correlated & high‑VIF features but PROTECTS specified indices."""
    def __init__(self, corr_threshold=0.7, vif_threshold=5, protect_indices=None):
        self.corr_threshold = corr_threshold
        self.vif_threshold = vif_threshold
        self.protect_indices = protect_indices if protect_indices is not None else []
        self.keep_indices_ = None

    def fit(self, X, y=None):
        X = np.asarray(X)
        n_features = X.shape[1]
        if n_features <= 1:
            self.keep_indices_ = np.arange(n_features)
            return self

        protected = set(self.protect_indices)
        corr = np.corrcoef(X, rowvar=False)
        high_corr_pairs = []
        for i in range(n_features):
            for j in range(i+1, n_features):
                if abs(corr[i, j]) > self.corr_threshold:
                    high_corr_pairs.append((i, j))
        mean_abs_corr = np.mean(np.abs(corr), axis=0)
        drop_set = set()
        for i, j in high_corr_pairs:
            if i in drop_set or j in drop_set:
                continue
            # If one is protected, drop the other
            if i in protected:
                drop_set.add(j)
            elif j in protected:
                drop_set.add(i)
            else:
                # both unprotected: drop the one with larger mean abs corr
                if mean_abs_corr[i] >= mean_abs_corr[j]:
                    drop_set.add(i)
                else:
                    drop_set.add(j)
        keep_indices = [i for i in range(n_features) if i not in drop_set]

        # VIF on unprotected features only
        unprotected_remaining = [i for i in keep_indices if i not in protected]
        if len(unprotected_remaining) >= 2:
            X_rem = X[:, unprotected_remaining]
            stds = np.std(X_rem, axis=0)
            non_const = [idx for idx, s in enumerate(stds) if s > 1e-10]
            if len(non_const) >= 2:
                X_vif = X_rem[:, non_const]
                vif_data = np.zeros(len(non_const))
                for idx in range(len(non_const)):
                    vif_data[idx] = variance_inflation_factor(X_vif, idx)
                vif_drop_rel = [non_const[i] for i, vif in enumerate(vif_data) if vif >= self.vif_threshold]
                vif_drop_abs = [unprotected_remaining[i] for i in vif_drop_rel]
                keep_indices = [i for i in keep_indices if i not in vif_drop_abs]

        self.keep_indices_ = np.array(keep_indices)
        return self

    def transform(self, X):
        X = np.asarray(X)
        return X[:, self.keep_indices_]


class RFSelectorKeepCountries(BaseEstimator, TransformerMixin):
    """RF‑based feature selection that ALWAYS keeps specified indices (e.g., countries)."""
    def __init__(self, n_estimators=100, random_state=42, n_jobs=-1, threshold='median', keep_indices=None):
        self.n_estimators = n_estimators
        self.random_state = random_state
        self.n_jobs = n_jobs
        self.threshold = threshold
        self.keep_indices = keep_indices if keep_indices is not None else []
        self.support_ = None
        self.feature_importances_ = None

    def fit(self, X, y):
        X = np.asarray(X)
        self.rf_ = RandomForestClassifier(n_estimators=self.n_estimators,
                                          random_state=self.random_state,
                                          n_jobs=self.n_jobs)
        self.rf_.fit(X, y)
        self.feature_importances_ = self.rf_.feature_importances_
        # Determine threshold for non‑protected features
        if self.threshold == 'median':
            thresh = np.median(self.feature_importances_)
        elif self.threshold == 'mean':
            thresh = np.mean(self.feature_importances_)
        else:
            thresh = float(self.threshold)
        # Keep if in keep_indices OR importance >= threshold
        keep_set = set(self.keep_indices)
        self.support_ = np.array([(i in keep_set) or (self.feature_importances_[i] >= thresh)
                                  for i in range(len(self.feature_importances_))])
        return self

    def transform(self, X):
        X = np.asarray(X)
        return X[:, self.support_]


class CleanCommunityTransformer(BaseEstimator, TransformerMixin):
    """Cleans community variables: strips '%', converts to numeric, and handles missing."""
    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()
        if isinstance(X, pd.DataFrame):
            for col in X.columns:
                if X[col].dtype == 'object':
                    X[col] = X[col].str.replace('%', '').str.strip()
                    X[col] = pd.to_numeric(X[col], errors='coerce')
        return X

    def get_feature_names_out(self, input_features=None):
        if input_features is None:
            return None
        return np.asarray(input_features, dtype=object)


# ============================================================
# 2. CATBOOST WRAPPER (accepts class_weights)
# ============================================================
class CatBoostSklearnWrapper(BaseEstimator, ClassifierMixin):
    def __init__(self, iterations=500, depth=6, learning_rate=0.05,
                 l2_leaf_reg=3.0, border_count=128, random_strength=1.0,
                 bagging_temperature=1.0, min_data_in_leaf=1,
                 auto_class_weights="Balanced", class_weights=None,
                 verbose=0, random_state=42):
        self.iterations = iterations
        self.depth = depth
        self.learning_rate = learning_rate
        self.l2_leaf_reg = l2_leaf_reg
        self.border_count = border_count
        self.random_strength = random_strength
        self.bagging_temperature = bagging_temperature
        self.min_data_in_leaf = min_data_in_leaf
        self.auto_class_weights = auto_class_weights
        self.class_weights = class_weights
        self.verbose = verbose
        self.random_state = random_state

    def _make_model(self):
        params = dict(
            iterations=self.iterations,
            depth=self.depth,
            learning_rate=self.learning_rate,
            l2_leaf_reg=self.l2_leaf_reg,
            border_count=self.border_count,
            random_strength=self.random_strength,
            bagging_temperature=self.bagging_temperature,
            min_data_in_leaf=self.min_data_in_leaf,
            verbose=self.verbose,
            random_seed=self.random_state,
        )
        if self.class_weights is not None:
            params['class_weights'] = self.class_weights
        elif self.auto_class_weights is not None:
            params['auto_class_weights'] = self.auto_class_weights
        return CatBoostClassifier(**params)

    def fit(self, X, y):
        self.model_ = self._make_model()
        self.model_.fit(X, y)
        self.classes_ = np.unique(y)
        return self

    def predict(self, X):
        return self.model_.predict(X).flatten().astype(int)

    def predict_proba(self, X):
        return self.model_.predict_proba(X)

# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================
def clean_columns(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
        .str.replace("-", "_", regex=False)
    )
    return df

def find_best_threshold(y_true, y_prob, thresholds=None):
    if thresholds is None:
        thresholds = np.arange(0.10, 0.95, 0.05)
    best_thr = 0.5
    best_f1 = -1
    for thr in thresholds:
        y_pred = (y_prob >= thr).astype(int)
        f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)
        if f1 > best_f1:
            best_f1 = f1
            best_thr = thr
    return best_thr

def get_proba(pipeline, X):
    try:
        proba = pipeline.predict_proba(X)
        if proba.ndim == 2 and proba.shape[1] == 2:
            return proba[:, 1]
        else:
            return pipeline.predict(X).astype(float)
    except AttributeError:
        if hasattr(pipeline, 'decision_function'):
            scores = pipeline.decision_function(X)
            proba = 1 / (1 + np.exp(-scores))
            if proba.ndim == 2:
                proba = proba[:, 1] if proba.shape[1] == 2 else proba.ravel()
            return proba
        else:
            return pipeline.predict(X).astype(float)

# ============================================================
# 4. LOAD RAW DATA
# ============================================================
train_raw = pd.read_csv(r"C:\Users\wub\desktop\train_data.csv", encoding="utf-8-sig")
test_raw  = pd.read_csv(r"C:\Users\wub\desktop\test_data.csv", encoding="utf-8-sig")

train_raw = clean_columns(train_raw)
test_raw  = clean_columns(test_raw)

TARGET = "water_access"
train_raw[TARGET] = train_raw[TARGET].map({"Basic_water_access": 0, "Limited_water_access": 1})
test_raw[TARGET]  = test_raw[TARGET].map({"Basic_water_access": 0, "Limited_water_access": 1})
train_raw = train_raw.dropna(subset=[TARGET])
test_raw  = test_raw.dropna(subset=[TARGET])

ID_COLS = ['case_identification', 'cluster_number', 'clusterid', 'household_sample_weight']
X_raw_train = train_raw.drop(columns=[TARGET] + ID_COLS, errors='ignore')
y_train = train_raw[TARGET]
X_raw_test  = test_raw.drop(columns=[TARGET] + ID_COLS, errors='ignore')
y_test = test_raw[TARGET]

# ============================================================
# 5. FEATURE LISTS AND CATEGORIES (exclude unwanted one‑hot categories)
# ============================================================
binary_vars = [
    "type_of_place_of_residence", "has_electricity", "sex_of_head_of_household",
    "household_water_treatment", "shared_toilets_with_others",
    "handwashing_place_observed", "number_of_households_members",
    "items_present_soap_or_detergent", "household_media_exposure",
    "sanitation_status", "housing_material_status"
]
onehot_vars = ["person_fetching_water", "current_marital_status", "country",
               "children_under5_group", "age_of_household_head"]
ordinal_vars = ["highest_educational_level_attained", "wealth_index_combined"]
community_vars = [
    "community_level_education", "community_level_electricity",
    "community_level_water_treatment", "community_level_media_exposure",
    "community_level_toilet", "community_level_poverty"
]

# ---- Determine categories for one‑hot variables ----
# Exclude "others" and "never married"
person_cats = sorted(X_raw_train["person_fetching_water"].dropna().unique())
marital_cats = sorted(X_raw_train["current_marital_status"].dropna().unique())
person_cats_keep = [c for c in person_cats if c.lower() != "others"]
marital_cats_keep = [c for c in marital_cats if c.lower() != "never married"]

# For the rest, keep all
country_cats = sorted(X_raw_train["country"].dropna().unique())
children_cats = sorted(X_raw_train["children_under5_group"].dropna().unique())
age_cats = sorted(X_raw_train["age_of_household_head"].dropna().unique())

onehot_categories = [
    person_cats_keep,
    marital_cats_keep,
    country_cats,
    children_cats,
    age_cats
]

# ============================================================
# 6. COLUMN TRANSFORMER (with custom one‑hot categories)
# ============================================================
binary_categories = {
    "type_of_place_of_residence": ["rural", "urban"],
    "has_electricity": ["no", "yes"],
    "sex_of_head_of_household": ["female", "male"],
    "household_water_treatment": ["no", "yes"],
    "shared_toilets_with_others": ["no", "yes"],
    "handwashing_place_observed": ["no", "yes"],
    "number_of_households_members": ["<4", ">=4"],
    "items_present_soap_or_detergent": ["no", "yes"],
    "household_media_exposure": ["No", "Yes"],
    "sanitation_status": ["unimproved_sanitation", "improved_sanitation"],
    "housing_material_status": ["unimproved_housing_material", "improved_housing_material"]
}
binary_categories_list = [binary_categories[col] for col in binary_vars]

binary_transformer = SkPipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(categories=binary_categories_list))
])

onehot_transformer = SkPipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(
        categories=onehot_categories,
        drop='first',
        handle_unknown='ignore',
        sparse_output=False
    ))
])

ordinal_categories = [
    ["No education", "Primary", "Secondary", "Higher"],
    ["Poorest", "Poorer", "Middle", "Richer", "Richest"]
]
ordinal_transformer = SkPipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(categories=ordinal_categories))
])

community_transformer = SkPipeline([
    ('clean_convert', CleanCommunityTransformer()),
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

numeric_transformer = SkPipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Determine numeric columns (all that are not in the special lists)
all_special = set(binary_vars + onehot_vars + ordinal_vars + community_vars)
numeric_cols = [col for col in X_raw_train.columns if col not in all_special]

preprocessor = ColumnTransformer([
    ('binary', binary_transformer, binary_vars),
    ('onehot', onehot_transformer, onehot_vars),
    ('ordinal', ordinal_transformer, ordinal_vars),
    ('community', community_transformer, community_vars),
    ('numeric', numeric_transformer, numeric_cols)
])

# ============================================================
# 7. IDENTIFY COUNTRY FEATURE INDICES (to protect)
# ============================================================
preprocessor.fit(X_raw_train.iloc[:5])
feature_names = preprocessor.get_feature_names_out()
country_indices = [i for i, name in enumerate(feature_names) if 'onehot__country_' in name]
print(f"Country features protected ({len(country_indices)}):")
for idx in country_indices:
    print(f"  {feature_names[idx]}")

# ============================================================
# 8. CLASS WEIGHTS
# ============================================================
counter = Counter(y_train)
majority = max(counter, key=counter.get)
minority = min(counter, key=counter.get)
scale_pos_weight = counter[majority] / counter[minority]
classes = np.unique(y_train)
class_weights_dict = {cls: w for cls, w in zip(classes, compute_class_weight('balanced', classes=classes, y=y_train))}
catboost_weights = [1, scale_pos_weight]

# ============================================================
# 9. MODELS (SVM changed to loss='log' for probability)
# ============================================================
base_models = {
    "LR": LogisticRegression(max_iter=2000, C=1.0, solver='lbfgs', random_state=42),
    "SVM": SGDClassifier(loss='log_loss', alpha=1e-4, max_iter=2000, tol=1e-3, random_state=42),
    "RF": RandomForestClassifier(n_estimators=300, max_depth=12, min_samples_split=5, min_samples_leaf=2, max_features='sqrt', n_jobs=-1, random_state=42),
    "DT": DecisionTreeClassifier(max_depth=10, min_samples_split=5, min_samples_leaf=2, random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=7, weights='distance', n_jobs=-1),
    "ANN": MLPClassifier(hidden_layer_sizes=(64,32), activation='relu', solver='adam', alpha=1e-4, learning_rate='adaptive', max_iter=500, early_stopping=True, n_iter_no_change=15, random_state=42),
    "XGB": XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0, eval_metric='logloss', random_state=42),
    "CatBoost": CatBoostSklearnWrapper(iterations=300, depth=6, learning_rate=0.05, l2_leaf_reg=3.0, verbose=0, random_state=42),
    "NB": GaussianNB(),
}

# ============================================================
# 10. BUILD PIPELINE (protects country features)
# ============================================================
def build_pipeline(classifier, imbalance_method):
    steps = [
        ('prep', preprocessor),
        ('corr_vif', CorrelationVIFSelector(
            corr_threshold=0.9, vif_threshold=10,
            protect_indices=country_indices
        )),
        ('rf_select', RFSelectorKeepCountries(
            n_estimators=100, random_state=42, n_jobs=-1,
            threshold='median',
            keep_indices=country_indices
        )),
    ]
    if imbalance_method == 'SMOTE':
        steps.append(('sampler', SMOTE(random_state=42)))
    elif imbalance_method == 'SMOTEENN':
        steps.append(('sampler', SMOTEENN(random_state=42)))
    steps.append(('clf', classifier))
    return ImbPipeline(steps)

# ============================================================
# 11. EVALUATION
# ============================================================
imbalance_methods = ['SMOTE', 'SMOTEENN', 'ClassWeight']
cv_outer = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = []
all_pipelines = []

print("\n" + "="*80)
print("STARTING EVALUATION")
print("="*80)

for model_name, clf in base_models.items():
    for method in imbalance_methods:
        print(f"\n--- {model_name} + {method} ---")

        if method == 'ClassWeight':
            clf_copy = clf.__class__(**clf.get_params())
            if model_name in ['LR', 'RF', 'DT', 'SVM']:
                clf_copy.set_params(class_weight='balanced')
            elif model_name == 'XGB':
                clf_copy.set_params(scale_pos_weight=scale_pos_weight)
            elif model_name == 'CatBoost':
                # Use the wrapper with explicit class_weights, not auto_class_weights
                clf_copy = CatBoostSklearnWrapper(
                    iterations=clf.get_params().get('iterations', 300),
                    depth=clf.get_params().get('depth', 6),
                    learning_rate=clf.get_params().get('learning_rate', 0.05),
                    l2_leaf_reg=clf.get_params().get('l2_leaf_reg', 3.0),
                    class_weights=catboost_weights,
                    auto_class_weights=None,   # important: don't pass both
                    verbose=0,
                    random_state=42
                )
            else:
                # For other models (e.g., NB, KNN, ANN) that don't support class_weight, we keep as is
                pass
            pipeline = build_pipeline(clf_copy, method)
        else:
            clf_copy = clf.__class__(**clf.get_params())
            pipeline = build_pipeline(clf_copy, method)

        # Manual CV loop for OOF probabilities
        oof_probs = np.zeros(len(y_train))
        for train_idx, val_idx in cv_outer.split(X_raw_train, y_train):
            X_tr, y_tr = X_raw_train.iloc[train_idx], y_train.iloc[train_idx]
            X_val = X_raw_train.iloc[val_idx]
            from sklearn.base import clone
            pipe_clone = clone(pipeline)
            pipe_clone.fit(X_tr, y_tr)
            oof_probs[val_idx] = get_proba(pipe_clone, X_val)

        best_thr = find_best_threshold(y_train, oof_probs)
        print(f"   Tuned threshold: {best_thr:.3f}")

        pipeline.fit(X_raw_train, y_train)

        y_prob_test = get_proba(pipeline, X_raw_test)
        y_pred_test = (y_prob_test >= best_thr).astype(int)

        acc = accuracy_score(y_test, y_pred_test)
        prec = precision_score(y_test, y_pred_test, average='weighted', zero_division=0)
        rec = recall_score(y_test, y_pred_test, average='weighted', zero_division=0)
        f1 = f1_score(y_test, y_pred_test, average='weighted', zero_division=0)
        roc = roc_auc_score(y_test, y_prob_test)
        ap = average_precision_score(y_test, y_prob_test)

        print(f"   Test: Acc={acc:.4f}, Prec={prec:.4f}, Rec={rec:.4f}, F1={f1:.4f}, ROC={roc:.4f}, AP={ap:.4f}")

        results.append({
            'Model': model_name,
            'Method': method,
            'Accuracy': acc,
            'Precision': prec,
            'Recall': rec,
            'F1': f1,
            'ROC_AUC': roc,
            'AP_AUC': ap,
            'Threshold': best_thr
        })
        all_pipelines.append((model_name, method, pipeline))

# ============================================================
# 12. SAVE RESULTS & BAR CHARTS
# ============================================================
output_path = r"C:\Users\wub\Desktop\ABC\ABC"
results_df = pd.DataFrame(results)
results_df.to_csv(f"{output_path}/comparison_results.csv", index=False)

metrics = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC_AUC', 'AP_AUC']
colors = {'SMOTE': '#1f77b4', 'SMOTEENN': '#ff7f0e', 'ClassWeight': '#d62728'}

fig, axes = plt.subplots(len(metrics), 1, figsize=(14, 11))
for i, metric in enumerate(metrics):
    ax = axes[i]
    sns.barplot(data=results_df, x='Model', y=metric, hue='Method', palette=colors, dodge=True, ax=ax)
    ax.set_ylabel(metric, fontsize=10)
    ax.set_ylim(0, 1.0)
    ax.tick_params(axis='x', rotation=30)
    ax.get_legend().remove()
    for p in ax.patches:
        height = p.get_height()
        if height > 0:
            ax.annotate(f'{height:.3f}', (p.get_x()+p.get_width()/2., height), ha='center', va='bottom', fontsize=7)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, title="Class imbalance Handling Method", loc="upper center", ncol=3)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.savefig(f"{output_path}/performance_comparison_methods.png", dpi=300, bbox_inches='tight')
plt.show()

# ============================================================
# 13. CORRELATION HEATMAP AND SELECTED FEATURES (best model)
# ============================================================
best_idx = results_df['ROC_AUC'].idxmax()
best_model_name = results_df.loc[best_idx, 'Model']
best_method = results_df.loc[best_idx, 'Method']
best_pipeline = None
for mn, meth, pipe in all_pipelines:
    if mn == best_model_name and meth == best_method:
        best_pipeline = pipe
        break

if best_pipeline is not None:
    print(f"\n🏆 Best model: {best_model_name} + {best_method} (ROC-AUC={results_df.loc[best_idx, 'ROC_AUC']:.4f})")

    # ---------- Correlation Heatmap ----------
    X_prep = best_pipeline.named_steps['prep'].transform(X_raw_train)
    try:
        feature_names = best_pipeline.named_steps['prep'].get_feature_names_out()
    except:
        feature_names = [f"feat_{i}" for i in range(X_prep.shape[1])]
    X_prep_df = pd.DataFrame(X_prep, columns=feature_names)

    corr = X_prep_df.corr()
    mask = np.triu(np.ones_like(corr, dtype=bool))
    plt.figure(figsize=(16, 12))
    sns.heatmap(corr, mask=mask, cmap='coolwarm', center=0,
                square=True, linewidths=0.5, annot=False,
                cbar_kws={"shrink": 0.8})
    plt.title(f"Correlation Heatmap (Preprocessed Training Data) - {best_model_name} + {best_method}")
    plt.tight_layout()
    plt.savefig(f"{output_path}/correlation_heatmap.png", dpi=300, bbox_inches='tight')
    plt.show()

    # ---------- Extract final selected features ----------
    corr_vif = best_pipeline.named_steps['corr_vif']
    rf_sel = best_pipeline.named_steps['rf_select']

    kept_indices = corr_vif.keep_indices_
    support = rf_sel.support_

    final_indices = kept_indices[support]
    final_feature_names = [feature_names[i] for i in final_indices]

    removed_corr_vif_indices = [i for i in range(len(feature_names)) if i not in kept_indices]
    removed_corr_vif_names = [feature_names[i] for i in removed_corr_vif_indices]

    rf_removed_indices = kept_indices[~support]
    rf_removed_names = [feature_names[i] for i in rf_removed_indices]

    selected_df = pd.DataFrame({'Selected_Features': final_feature_names})
    selected_df.to_csv(f"{output_path}/selected_features.csv", index=False)

    pd.DataFrame({'Removed_by_CorrVIF': removed_corr_vif_names}).to_csv(
        f"{output_path}/removed_by_corr_vif.csv", index=False)
    pd.DataFrame({'Removed_by_RF': rf_removed_names}).to_csv(
        f"{output_path}/removed_by_rf.csv", index=False)

    print(f"\n✅ Final selected features ({len(final_feature_names)}) saved to {output_path}/selected_features.csv")
    print(f"✅ Removed features saved to {output_path}/removed_by_corr_vif.csv and {output_path}/removed_by_rf.csv")
    print(f"✅ Correlation heatmap saved to {output_path}/correlation_heatmap.png")

else:
    print("No pipeline found for the best model.")

print(f"\n✅ All comparison results saved to {output_path}")
```
